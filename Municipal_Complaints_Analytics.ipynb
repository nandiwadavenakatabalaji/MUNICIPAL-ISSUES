```python
import pandas as pd

df = pd.read_csv('/content/municipal_training_set_1100.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

# 1. Column names and data types
print("COLUMN INFORMATION:")
df.info()

# 2. Missing values check
print("\nMISSING VALUES:")
print(df.isnull().sum())

# 3. Duplicate rows check
print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())

# 4. Dataset summary
print("\nDATASET SUMMARY:")
display(df.describe(include="all").T)
```

```python

# Step 11: Data Cleaning

# Make a copy of the original dataset
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Check missing values
print("Missing values before cleaning:")
print(df_clean.isnull().sum())

# Display cleaned dataset information
print("\nDataset shape after removing duplicates:")
print(df_clean.shape)

# Display first 5 rows
display(df_clean.head())
```

```python

print("Available columns:")
print(df_clean.columns.tolist())
```

```python

# Convert date columns into datetime format
df_clean['report_datetime'] = pd.to_datetime(
    df_clean['report_datetime'], errors='coerce'
)

df_clean['resolution_datetime'] = pd.to_datetime(
    df_clean['resolution_datetime'], errors='coerce'
)

# Calculate resolution time in hours
df_clean['resolution_hours'] = (
    df_clean['resolution_datetime'] -
    df_clean['report_datetime']
).dt.total_seconds() / 3600

# Keep valid resolution times only
resolved = df_clean[
    (df_clean['resolution_hours'] >= 0) &
    (df_clean['resolution_datetime'].notna())
]

print("Total complaints:", len(df_clean))
print("Complaints with valid resolution time:", len(resolved))

if len(resolved) > 0:
    print(
        "Average resolution time (hours):",
        round(resolved['resolution_hours'].mean(), 2)
    )
else:
    print("No valid resolution times found.")
```

```python

# Count complaints by priority level
priority_counts = df_clean['priority_level'].value_counts()

print("Complaints by Priority Level:")
display(priority_counts.to_frame('Total Complaints'))

# Create a bar chart
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
priority_counts.plot(kind='bar')
plt.title('Complaints by Priority Level')
plt.xlabel('Priority Level')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

city_counts = df_clean['location_city'].value_counts().head(10)

print("Top 10 Cities by Complaint Count:")
display(city_counts.to_frame('Total Complaints'))

plt.figure(figsize=(10, 5))
city_counts.plot(kind='bar')
plt.title('Top 10 Cities by Municipal Complaints')
plt.xlabel('City')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

# Count complaints by department
department_counts = df_clean['municipal_department'].value_counts()

print("Department-wise Complaint Count:")
display(department_counts.to_frame('Total Complaints'))

# Create chart
plt.figure(figsize=(10, 5))
department_counts.plot(kind='bar')
plt.title('Municipal Complaints by Department')
plt.xlabel('Municipal Department')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python

status_counts = df_clean['issue_status'].value_counts()

print("Complaint Status Summary:")
display(status_counts.to_frame('Total Complaints'))

plt.figure(figsize=(8, 5))
status_counts.plot(kind='bar')
plt.title('Municipal Complaint Status')
plt.xlabel('Complaint Status')
plt.ylabel('Number of Complaints')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
